```python

# 当前 Q 值
q_old = 4.0

# 本次交互获得的奖励
reward = 2.0

# 下一状态各动作的 Q 值
next_q_values = [3.0, 8.0, 5.0]

# 参数
gamma = 0.9
alpha = 0.5

# 下一状态的最大 Q 值
best_next_q = max(next_q_values)

# 计算目标值
target = reward + gamma * best_next_q

# 计算 TD 误差
td_error = target - q_old

# 更新 Q 值
q_new = q_old + alpha * td_error

print("目标值:", target)
print("TD误差:", td_error)
print("更新后的Q值:", q_new)
```
