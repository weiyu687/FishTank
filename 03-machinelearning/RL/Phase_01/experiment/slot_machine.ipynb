前面的内容已经认识了强化学习的基本概念、奖励与回报，以及智能体如何在**探索**和**利用**之间选择。

现在，把这些知识变成真正能允许的程序。

**该实验的目标是让一个原本不知道哪台老虎机最赚钱的智能体，通过不断尝试，逐渐学会选择收益最高的老虎机。**

首先需要明确需要解决的问题。

假如现在有三台老虎机A、B、C，其真实平均奖励如下表所示：

|老虎机|真实平均奖励|
|:-:|:-:|
|A|2|
|B|5|
|C|8|

但智能体并不知道上述信息。

其只能通过拉动老虎机，观察每次得到的奖励，再决定下一步选择哪台。

然后我们对问题进行分析。

智能体需要记住两个信息：
- Q：每台老虎机目前估计的平均奖励。
- N：每台老虎机已经被选择了多少次。

例如，假如智能体选择老虎机B三次，得到的奖励是：

第一次 4

第二次 7

第三次 4

那么智能体对B的奖励估计是：
$$
Q(B)=\frac{4+7+4}{3}=5
$$
每得到一次新奖励，就重新计算平均值。

在强化学习中，可以用下面的更新公式：
$$
Q_{new}=Q_{old}+\frac{1}{N}(R-Q_{old})
$$
其中：
- $Q_{old}$：更新前的平均奖励估计
- $R$：本次获得的奖励
- $N$：目前总共选择该老虎机的次数
- $Q_{new}$：更新后的估计
上述公式是后续学习Q-learning的基础之一。

接下来开始程序的实现。

```python
import random

# 1. 三台老虎机的真实平均奖励
true_rewards = [2, 5, 8]

# 2. 智能体的初始知识
Q = [0.0, 0.0, 0.0]
N = [0, 0, 0]

# 3. 设置探索概率
epsilon = 0.1

# 4. 训练 1000 次
for step in range(1000):

    # 探索： 随机选择一台老虎机
    if random.random() < epsilon:
        action = random.randint(0, 2)

    # 利用： 选择目前估计奖励最高的老虎机
    else:
        action = Q.index(max(Q))

    # 5. 获得奖励（带有随机波动）
    reward = true_rewards[action] + random.uniform(-2, 2)

    # 6. 更新选择次数
    N[action] += 1

    # 7. 更新奖励估计
    Q[action] += (reward - Q[action]) / N[action]

# 8. 输出训练结果
print("每台老虎机被选择的次数：", N)
print("每台老虎机的估计奖励：", [round(q, 2) for q in Q])
print("智能体认为最好的老虎机：", Q.index(max(Q)) + 1)
```

可以补充以下三个实验：
1. 将$\epsilon$设置为0%，观察智能体会不会陷入错误选择。
2. 将$\epsilon$设置为10%，比较训练 100 次和 5,000 次的结果。
3. 将$\epsilon$设置为100%，思考：即使智能体已经知道 C 最好，它还会不会优先选择 C？
